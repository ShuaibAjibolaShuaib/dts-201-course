```python
import pandas as pd

# 1. Load the CSV file
df = pd.read_csv("50_startups.csv")

# Display the number of rows and columns
print("1. Shape of the dataset:")
print(df.shape)
print()


# 2. Display information about the dataset
print("2. Dataset information:")
df.info()
print()


# 3. Check for missing values
print("3. Missing values:")
print(df.isnull().sum())
print()

# Check for "Unknown" values
print("Invalid/Unknown values:")
print((df == "Unknown").sum())
print()

# Convert Marketing Spend to numbers
# Invalid values such as "Unknown" become NaN
df["Marketing Spend"] = pd.to_numeric(
    df["Marketing Spend"],
    errors="coerce"
)

print("Missing values after cleaning:")
print(df.isnull().sum())
print()


# 4. Create a Series and a single-column DataFrame
series = df["State"]
dataframe = df[["State"]]

print("4. Series:")
print(type(series))
print("Shape:", series.shape)
print()

print("Single-column DataFrame:")
print(type(dataframe))
print("Shape:", dataframe.shape)
print()


# 5. Display descriptive statistics
print("5. Descriptive statistics:")
print(df.describe())
print()

# Calculate the mean Profit
print("Mean Profit:")
print(df["Profit"].mean())
print()


# 6. Export the DataFrame to JSON
df.to_json("50_startups.json", orient="records")

print("6. JSON file created successfully.")
print()

# Open and display the JSON file
with open("50_startups.json", "r") as file:
    json_data = file.read()

print("JSON file contents:")
print(json_data)
print()


# 7. Examples of data that would not help predict a student's grade
print("7. Examples:")
print("1. Student's favorite color")
print("2. Student's favorite food")
print("3. Student's phone brand")
```
