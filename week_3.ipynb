```python
import pandas as pd

# 1. Load dataset
df = pd.read_excel("50_Startups_messy.xlsx")

# Report shape, missing values and duplicates
print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate count:")
print(df.duplicated().sum())


# 2. Clean the data

# Drop duplicates
clean = df.drop_duplicates().copy()

# Standardise State
clean["State"] = clean["State"].str.strip().str.title()

# Convert Marketing Spend to numeric
clean["Marketing Spend"] = pd.to_numeric(
    clean["Marketing Spend"],
    errors="coerce"
)

print("\nAfter cleaning:")
print(clean.isnull().sum())

print("\nState counts:")
print(clean["State"].value_counts())


# 3. Simple random sample of 20
simple_sample = clean.sample(
    n=20,
    random_state=7
)

print("\nSimple random sample:")
print(simple_sample)

print("\nSimple sample State counts:")
print(simple_sample["State"].value_counts())


# 4. Stratified 20% sample by State
stratified_sample = (
    clean.groupby("State", group_keys=False)
    .sample(frac=0.20, random_state=7)
)

print("\nStratified sample:")
print(stratified_sample)

print("\nStratified sample State counts:")
print(stratified_sample["State"].value_counts())


# 5. Calculate means

columns = ["R&D Spend", "Marketing Spend"]

population_mean = clean[columns].mean()
simple_mean = simple_sample[columns].mean()
stratified_mean = stratified_sample[columns].mean()

print("\nPopulation means:")
print(population_mean)

print("\nSimple random sample means:")
print(simple_mean)

print("\nStratified sample means:")
print(stratified_mean)


# 6. Compare differences

simple_difference = abs(simple_mean - population_mean)
stratified_difference = abs(stratified_mean - population_mean)

print("\nSimple random differences:")
print(simple_difference)

print("\nStratified differences:")
print(stratified_difference)
```
